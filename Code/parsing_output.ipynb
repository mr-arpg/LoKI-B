# NH3

***

### spatial coefficients:

- ionization ($ \frac{\alpha}{N} $)
- attachment ($ \frac{\eta}{N} $)

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

# Create the folder "imgs" if it doesn't exist
output_folder = "imgs"
os.makedirs(output_folder, exist_ok=True)

# Load the simulated data from the text file
simulated_file_path = "LoKI-B/Output/swarm_NH3_spatial/lookUpTableSwarm.txt"
simulated_data = pd.read_csv(simulated_file_path, delim_whitespace=True)

# Extract the relevant simulated data columns
red_field_sim = simulated_data.filter(like="RedField").iloc[:, 0]
red_townsend_sim = simulated_data.filter(like="RedTow").iloc[:, 0]
red_attachment_sim = simulated_data.filter(like="RedAtt").iloc[:, 0]

# Load the measured data from the Excel file
measured_file_path = "Measured swarm parameters for comparison.xlsx"
measured_data = pd.read_excel(measured_file_path, sheet_name="NH3")

# Extract measured values for Townsend
red_field_meas_townsend = measured_data.iloc[5:14, 12]  # Column M
red_townsend_meas = measured_data.iloc[5:14, 13]        # Column N

# Extract measured values for Attachment
red_field_meas_attachment = measured_data.iloc[5:16, 17]  # Column R
red_attachment_meas = measured_data.iloc[5:16, 18]        # Column S

# Plot Townsend (Ionization Coefficients)
plt.figure(figsize=(10, 6))
plt.plot(red_field_sim, red_townsend_sim, "-", label="Simulated Data (Trend)", color="blue", linewidth=2)  # Simulated as a line
plt.plot(red_field_meas_townsend, red_townsend_meas, "o", label="Measured Data (Points)", color="black")   # Measured as points
plt.xscale("log")
plt.yscale("log")
plt.xlim(10, 1e3)  # Set x-axis limits
plt.ylim(1e-26, None)  # Set y-axis lower limit to 10^-25, and let the upper limit be determined automatically
plt.xlabel("Reduced Field (Td)", fontsize=12)
plt.ylabel("Reduced Townsend Coefficient (m²)", fontsize=12)
plt.title("Ionization Coefficient vs Reduced Field", fontsize=14)
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()
plt.show()

# Plot Attachment Coefficients
plt.figure(figsize=(10, 6))
plt.plot(red_field_sim, red_attachment_sim, "-", label="Simulated Data (Trend)", color="red", linewidth=2)  # Simulated as a line
plt.plot(red_field_meas_attachment, red_attachment_meas, "o", label="Measured Data (Points)", color="black")  # Measured as points
plt.xscale("log")
plt.yscale("log")
plt.xlim(10, 1e2)
plt.ylim(1e-25, None)  # Set y-axis lower limit to 10^-25, and let the upper limit be determined automatically
plt.xlabel("Reduced Field (Td)", fontsize=12)
plt.ylabel("Reduced Attachment Coefficient (m²)", fontsize=12)
plt.title("Reduced Field vs Attachment Coefficient", fontsize=14)
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()
plt.show()

```

spatial coefficients

- dissociation coeff. ($ \alpha_{diss}/N$)
- emission coeff. ($ \alpha_{emiss}/N$)
- townsend ($ \frac{\alpha}{N}$)

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load simulated data from lookUpTableRateCoeff.txt
file_sim = "LoKI-B/Output/swarm_H2_spatial/lookUpTableRateCoeff.txt"
sim_file_path_h2 = "LoKI-B/Output/swarm_H2_spatial/lookUpTableSwarm.txt"
simulated_data_coeff = pd.read_csv(file_sim, delim_whitespace=True, comment="#")  # Adjust delimiter if needed
sim_data_h2 = pd.read_csv(sim_file_path_h2, delim_whitespace=True)
# Load drift velocity from the previous simulation
drift_velocity_sim_h2 = sim_data_h2.filter(like="DriftVelocity").iloc[:, 0]  # Ensure drift velocity data is available

# Extract relevant columns for reduced field and reaction rates
R9_sim_h2 = simulated_data_coeff.filter(like="R9").iloc[:, 0]  # R9 for dissociation
R5_sim_h2 = simulated_data_coeff.filter(like="R5").iloc[:, 0]  # R5 for emission
R7_sim_h2 = simulated_data_coeff.filter(like="R7").iloc[:, 0]  # R7 for emission
R11_sim_h2 = simulated_data_coeff.filter(like="R11").iloc[:, 0]  # R11 for emission
R21_sim_h2 = simulated_data_coeff.filter(like="R21").iloc[:, 0]  # R21 for Townsend
R22_sim_h2 = simulated_data_coeff.filter(like="R22").iloc[:, 0]  # R22 for Townsend

# Calculate coefficients
emission_coeff_sim_h2 = R9_sim_h2 / drift_velocity_sim_h2
dissociation_coeff_sim_h2 = (R5_sim_h2 + R7_sim_h2 + R11_sim_h2) / drift_velocity_sim_h2
ionization_coeff_sim_h2 = (R21_sim_h2 + R22_sim_h2) / drift_velocity_sim_h2

# Extract measured dissociation data
red_field_meas_dissociation = measured_data_h2.iloc[6:19, 17]  # Column R
dissociation_coeff_meas = measured_data_h2.iloc[6:19, 18]  # Column S

# Extract measured emission data
red_field_meas_emission = measured_data_h2.iloc[6:35, 22]  # Column W
emission_coeff_meas = measured_data_h2.iloc[6:35, 23]  # Column X

# Extract measured ionization data
red_field_meas_ionization = measured_data_h2.iloc[6:116, 12]  # Column M
ionization_coeff_meas = measured_data_h2.iloc[6:116, 13]  # Column N

# Plot Dissociation Coefficient
plt.figure(figsize=(10, 6))
plt.plot(red_field_sim_h2, dissociation_coeff_sim_h2, "-", label="Simulated Data (Trend)", color="blue", linewidth=2)
plt.plot(red_field_meas_dissociation, dissociation_coeff_meas, "o", label="Measured Data (Points)", color="black", markersize = 4)
plt.xscale("log")
plt.yscale("log")
plt.xlim(10, None)
plt.ylim(1e-25, None)
plt.xlabel("Reduced Field (Td)", fontsize=12)
plt.ylabel("Dissociation Coefficient (m³/s)", fontsize=12)
plt.title("Reduced Field vs Dissociation Coefficient (H2)", fontsize=14)
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()
plt.show()

# Plot Emission Coefficient
plt.figure(figsize=(10, 6))
plt.plot(red_field_sim_h2, emission_coeff_sim_h2, "-", label="Simulated Data (Trend)", color="red", linewidth=2)
plt.plot(red_field_meas_emission, emission_coeff_meas, "o", label="Measured Data (Points)", color="black", markersize = 3)
plt.xscale("log")
plt.yscale("log")
plt.xlim(10, None)
plt.ylim(1e-23, None)
plt.xlabel("Reduced Field (Td)", fontsize=12)
plt.ylabel("Emission Coefficient (m³/s)", fontsize=12)
plt.title("Reduced Field vs Emission Coefficient (H2)", fontsize=14)
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()
plt.show()

# Plot Ionization Coefficient (Townsend)
plt.figure(figsize=(10, 6))
plt.plot(red_field_sim_h2, ionization_coeff_sim_h2, "-", label="Simulated Data (Trend)", color="green", linewidth=2)
plt.plot(red_field_meas_ionization, ionization_coeff_meas, "o", label="Measured Data (Points)", color="black", markersize = 2)
plt.xscale("log")
plt.yscale("log")
plt.xlim(30, None)
plt.ylim(1e-25, None)
plt.xlabel("Reduced Field (Td)", fontsize=12)
plt.ylabel("Ionization Coefficient (m²)", fontsize=12)
plt.title("Ionization Coefficient vs Reduced Field (H2)", fontsize=14)
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()
plt.show()

```
